```python
# *args
def even_numbers(*args):
    for num in args:
        if num%2==0:
            print(f"the even numbers is{num}")
        else:
            print(f"the odd numbers is{num}")   
```

```python
even_numbers(1,45,78,56,3,17,90)
```

```python
#**kwargs
def brand_info(**kwargs):
    for (a,b) in kwargs.items():
        print(f"{a}:{b}")   
```

```python
brand_info(name="breeza",milage=600,type="4by2")
```

```python
#we can use both*args and **kwargs simentaneously
def result_info(*args,**kwargs):
    sum=0
    for num in args:
        sum+=num
    print(f"the total marks are {sum}")    
    for (a,b) in kwargs.items():
        print(f"{a}:{b}")        

```

```python
result_info(45,68,92,43,name="anam",section=112,school="gvei")
```
