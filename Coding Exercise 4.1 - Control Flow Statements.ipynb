```python
# Given a student's score, print their grade using the following criteria:

# 90 and above: "A"
# 80-89: "B"
# 70-79: "C"
# 60-69: "D"
# Below 60: "F"
num=56
if num>=90:
    print("A")
elif (num>=80) and (num<=89):
    print("B")
elif (num>=70) and (num<=79):
    print("C")
elif (num>=60) and (num<=69):
    print("D")
elif num<60:
    print("F")
else:
    print("fail")
```

```python
# Take the lengths of three sides of a triangle and determine its type:

# Equilateral (all sides are equal)
# Isosceles (two sides are equal)
# Scalene (all sides are different)
ab=10
bc=1
ca=15
if (ab==bc) and (ca==ab) and (bc==ca):
    print("equilateral triangle")
elif (ab==bc) or (ca==bc) or(ca==ab):
    print("isosceles triangle")
elif (ab!=bc) and (bc!=ca) and (ca!=ab):
    print("scalene triangle")
```
