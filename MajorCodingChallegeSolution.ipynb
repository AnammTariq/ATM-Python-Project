```python
def get_marks(n):
  marks=[]

  for i in range(n):
    mark=int(input(f"Enter marks of student:{i+1}"))
    marks.append(mark)

  return(marks)
```

```python
def calculate_statistics(m):
  total_marks=sum(m)
  average_marks=sum(m)/len(m)
  highest_marks=max(m)
  lowest_marks=min(m)

  return (total_marks,average_marks,highest_marks,lowest_marks)
```

```python
def get_grade(avg):
  if avg>=90:
    return("A")
  elif avg>=75 and avg<=89:
    return("B")
  elif avg>=60 and avg<=74:
    return("C")
  elif avg>=40 and avg<59:
    return ("D")
  else:
    return("F")
```

```python
def main():
  number_of_students=int(input("Enter the number of students"))
  marks=get_marks(number_of_students)

  tot_marks,avg_marks,high_marks,low_marks=calculate_statistics(marks)

  grade=get_grade(avg_marks)



  print(f"The list of marks is: {marks}")

  print(f"The total marks are: {tot_marks}")

  print(f"The average marks are: {avg_marks}")

  print(f"The highest and lowest marks are: {high_marks} & {low_marks} respectively")

  print(f"The overall grade is: {grade}")
```

```python
main()
```
