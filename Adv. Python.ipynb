```python

# Create a function called highest_score(*marks).
#
# It should accept any number of marks and return the highest mark.
#
# Test the function using marks from at least 3 students.

# Write your code below.
def highest_score(*marks):
    max_marks=0
    for num in marks:
        if num>max_marks:
            max_marks=num
    print(f"the highest marks is:{max_marks}")            
```

```python
highest_score(100,200,89)
```

```python

# Create a function called safe_divide(a, b).
#
# Divide a by b.
# Handle ZeroDivisionError if b is 0.
# Also handle TypeError if non-numeric values are supplied.
#
# Test:
# safe_divide(100, 5)
# safe_divide(100, 0)
# safe_divide("100", 5)

# Write your code below.
def safe_divide(a,b):
   try:
      print(a/b)
   except ZeroDivisionError: 
      print("you are dividing by zero")
   except TypeError:
      print("you wrote a non-numeric value")      
```

```python
safe_divide(100,5)
safe_divide(100,0)
safe_divide("100",5)
```
