```python
# Write a function sum_list() that takes a list of numbers as input and returns their sum.
def sum_list(l):
    sum=0
    for num in l:
        sum=sum+num
    print(sum)
```

```python
sum_list([1,2,3,4,5])
```

```python
# Write a function check_even_odd(num) that checks whether a given number is even or odd.
def check_even_odd(num):
    if num%2==0:
        print(f"the number {num} is even")
    else:
        print(f"the number {num} is odd")   
```

```python
# Write a function count_vowels(s) that takes a string and returns the count of vowels (a, e, i, o, u) in it.
def count_vowels(s):
    count=0
    vowels="a,e,i,o,u,A,E,I,O,U"
    for letter in s:
        if letter in vowels:
            count+=1
    print(count)    
```

```python
count_vowels("basket of apples")
```

```python
# Write a function max_in_tuple(t) that takes a tuple of numbers and returns the maximum number.
def max_in_tuple(t):
    max_number=0
    for num in t:
        if num>max_number:
            max_number=num
    print(f"the largest number is {max_number}")
    
```

```python
max_in_tuple((1,2,4,22,14))
```

```python
# Write a function key_exists(d, key) that checks if a given key exists in a dictionary.
def key_exists(d, key):
    if key in d:
        print(f"key {key} exists in this dictionary")
    else:
        print(f"key {key} does not exists in this dictionary") 
      
```

```python
d={"apple":50,"cherry":20,"kiwi":60}
key_exists(d, "apple")
    
```

```python
# Write a function set_intersection(set1, set2) that returns the intersection of two sets.
#intersection is common or matching elements between set1 and set2 (you have to print those only)
def set_intersection(set1, set2):
    for num in set1:
        if num in set2:
            print(num)
            
```

```python
set_intersection({1,2,3,4,5,5,3},{12,4,3,5,6,5})
```

```python
# Write a function word_count(sentence) that takes a sentence (string) and returns a dictionary where keys are words and values are their counts.
def count_in_sentence(s):
    letter=s.split()
    t=tuple(letter)
    d={}
    for word in t:
        d[word]=t.count(word)
    print(d)

    
```

```python
count_in_sentence("my name is anam tariq and my city is sxr") 
```
