```python
#Write a program that counts and prints the number of vowels in the string "machine learning".
s="machine learning"
count=0
vowels="a,e,i,o,u,A,E,I,O,U"
for letter in s:
    if letter in vowels:
     count+=1
print(count)
```

```python
# the given list is [1,4,6,7,9,35,18,13,20,45]
l=[1,4,6,7,9,35,18,13,20,45]
even=list()
odd=list()
for num in l:
   if num%2==0:
     even.append(num) 
   else:
      odd.append(num)
print(f"the odd list is:{odd}")
print(f"the even list is:{even}") 
```

```python
#given the list=["Abrar","mir","Srinagar","pune","delhi","is","why","name","age"]
l=["abrar","mir","srinagar","pune","delhi","is","why","name","age"]
name=list()
for char in l:
   if char in l[0:10:2]:
       name.append(char.capitalize())
   else:
       name.append(char)
print(name)
```

```python
#given the list=["Abrar","mir","Srinagar","pune","delhi","is","why","name","age"] {even}
l=["abrar","mir","srinagar","pune","delhi","is","why","name","age"]
name=list()
for char in l:
    if char in l[0::2]:
        name.append(char.capitalize())
    else:
        name.append(char)
print(name)            
```

```python
#l=["apple", "bike", "car" "chair", "lion", "elephant", "mansha", "house"] o/p more than 4 leeters in an element
l=["apple", "bike", "car", "house", "lion", "elephant", "mansha", "cat", "toy", "bat"]
l1=list()
for letter in l:
    if len(letter)>=4:
        l1.append(letter)
print(l1)              
```

```python
#l1=["apple", "bike", "doctor" "chair", "elephant", "mansha", "house"] and l2=["mohsin", "hand", "telivision", "eyes", "matches", "tab",]
l1=["apple", "bike", "doctor", "chair", "elephant", "mansha", "house"]
l2=["mohsin", "handslh", "telivision", "eyes", "matches", "tab",]
merge=l1+l2
l3=list()
for letter in merge:
    if len(letter)>=5:
        l3.append(letter.upper())
    else:
        l3.append(letter.capitalize())
print(l3)          
```

```python
#given a tuple in a list employee_list=[("anam",36),("mansha",28),("tehreem",44)] find the best employee with maximum working hours 
employee_list = [("anam", 40), ("mansha", 38), ("mohsin", 42), ("tehreem", 44)]
x=("anam",40)
y=("mansha",38)
p=("mohsin",42)
q=("tehreem",44)
if x[1] > y[1] and x[1] > p[1] and x[1] > q[1]:
    print("the best employee is {0} with maximum working hours {1}".format(x[0], x[1])) 
elif y[1] > x[1] and y[1] > p[1] and y[1] > q[1]:
    print("the best employee is {0} with maximum working hours {1}".format(y[0], y[1]))
elif p[1] > x[1] and p[1] > y[1] and p[1] > q[1]:
    print("the best employee is {0} with maximum working hours {1}".format(p[0], p[1]))
elif q[1] > x[1] and q[1] > y[1] and q[1] > p[1]:
    print("the best employee is {0} with maximum working hours {1}".format(q[0], q[1]))       
```
