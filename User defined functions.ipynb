```python
# Write a function to check if a given string is a palindrome. The function should take a string 
# as input and return True if it's a palindrome, otherwise False.
#(string is said to be palindrome if it is same in reverse order also e.g: civic, radar etc)
def palindrome_string(s):
    if s==s[ : :-1]:
      print(f"yes, {s} is palindrome")
    else:
       print(f"no, {s} is not palindrome")  
```

```python
palindrome_string("anam")
```

```python
# Write a function that takes a list of integers, sorts it in 
# ascending order, and then reverses the sorted list.
def ascending_and_reverse_list(l):
    l.sort()
    l.reverse()
    print(l)
    
```

```python
# Write a function to find the maximum and minimum of a list without using the max and min functions.
def max_min(l):
    max_number=0
    for num in l:
        if num>max_number:
            max_number=num
    min_number=max_number        
    for num in l:        
        if num<min_number:
            min_number=num
    print(f"the largest number is {max_number}")
    print(f"the smallest number is {min_number}")        
```

```python
max_min([1,23,45,89,56,23,90,12,34,4])
```

```python
# Write a function that takes a string and returns the count of vowels and consonants.
def vowels_consonants(s):
    vowel_count=0
    consonants_count=0
    vowels=("a,e,i,o,u,A,E,I,O,U")
    for letter in s:
        if letter in vowels:
            vowel_count+=1
        else:
           consonants_count+=1 
    print(f"the count of vowels is {vowel_count}")   
    print(f"the count of vowels is {consonants_count}")      
```

```python
vowels_consonants("anam tariq")
```

```python
# Write a function that takes a list of integers and returns two separate lists: one with even numbers and the other with odd numbers.
def even_odd(l):
    even_list=[]
    odd_list=[]
    for num in l:
        if num%2==0:
            even_list.append(num)
        else:
            odd_list.append(num) 
    print(f"the even list is {even_list}")   
    print(f"the odd list is {odd_list}")
    
```

```python
even_odd([12,34,2,78,45,33,90,67,24,8])      
```

```python
# Write a function that takes a sentence as input and returns 
# the count of each word in the sentence as a dictionary.
def word_count(s):
    letter=s.split()
    d={}
    for word in letter:
        d[word]=len(word)
    print(d)    
```

```python
word_count("mansha is buzy with her phone")
```
